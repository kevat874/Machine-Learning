### 2. Regression

```python
import os
os.environ["PATH"] = "/opt/homebrew/bin:" + os.environ["PATH"]

import pandas as pd
from sklearn import tree
from sklearn.datasets import fetch_california_housing
from dtreeviz import model

# Load California Housing dataset
boston = fetch_california_housing()

X_train = boston.data
y_train = boston.target

# Train Decision Tree Regressor
regr = tree.DecisionTreeRegressor(max_depth=1)
regr.fit(X_train, y_train)

# Visualize Decision Tree
viz = model(
    regr,
    X_train,
    y_train,
    target_name="price",
    feature_names=boston.feature_names
)

viz.view()
```

### 3. Horizontal Decision Tree

```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from dtreeviz import model

# Load Iris dataset
iris = load_iris()
X_train = iris.data
y_train = iris.target

# Train classifier
clas = DecisionTreeClassifier(max_depth=3, random_state=42)
clas.fit(X_train, y_train)

# Visualize decision tree
viz = model(
    clas,
    X_train,
    y_train,
    target_name='flower',
    feature_names=iris.feature_names,
    class_names=["setosa", "versicolor", "virginica"]
)

viz.view()
```

### 4. Show prediction path

```python
import numpy as np
from sklearn import tree
from sklearn.datasets import load_iris
from dtreeviz import model

clas = tree.DecisionTreeClassifier(random_state=42)
iris = load_iris()

X_train = iris.data
y_train = iris.target

clas.fit(X_train, y_train)

# Select a random sample
X = iris.data[np.random.randint(0, len(iris.data)), :]

# Visualize decision tree with selected sample
viz = model(
    clas,
    X_train,
    y_train,
    feature_names=iris.feature_names,
    class_names=["setosa", "versicolor", "virginica"]
)

viz.view(
    x=X
)
```

### 7. Show just prediction path, nothing else

```python
import numpy as np
from sklearn import tree
from sklearn.datasets import load_iris
from dtreeviz import model

clas = tree.DecisionTreeClassifier()
iris = load_iris()

X_train = iris.data
y_train = iris.target

clas.fit(X_train, y_train)

# Select a random Iris sample
X = iris.data[np.random.randint(0, len(iris.data)), :]

# Create visualization
viz = model(
    clas,
    X_train,
    y_train,
    target_name='flower',
    feature_names=iris.feature_names,
    class_names=["setosa", "versicolor", "virginica"]
)

# Show only the selected sample's decision path
viz.view(
    x=X,
    show_just_path=True
)
```

### 8. Prediction Path in Plain english

```python
from dtreeviz import explain_prediction_path

print(
    explain_prediction_path(
        clas,
        X,
        feature_names=iris.feature_names,
        explanation_type="plain_english"
    )
)
```
