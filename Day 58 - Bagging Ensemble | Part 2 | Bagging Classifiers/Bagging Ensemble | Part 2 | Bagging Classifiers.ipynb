```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

print("Libraries imported successfully")
```

```python
X, y = make_classification(n_samples=1000, n_features=6, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print("Data created. Train shape:", X_train.shape)
```

```python
dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)
print("Decision Tree Accuracy:", accuracy_score(y_test, dt.predict(X_test)))
```

```python
bag = BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=20,
    max_samples=0.5,
    random_state=42
)

bag.fit(X_train, y_train)
print("Bagging Accuracy:", accuracy_score(y_test, bag.predict(X_test)))
```

```python
pasting = BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=20,
    max_samples=0.5,
    bootstrap=False,
    random_state=42
)

pasting.fit(X_train, y_train)
print("Pasting Accuracy:", accuracy_score(y_test, pasting.predict(X_test)))
```

```python
subspaces = BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=20,
    max_samples=1.0,
    max_features=0.5,
    bootstrap_features=True,
    random_state=42
)

subspaces.fit(X_train, y_train)
print("Random Subspaces Accuracy:", accuracy_score(y_test, subspaces.predict(X_test)))
```

```python
patches = BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=20,
    max_samples=0.5,
    max_features=0.5,
    bootstrap=True,
    bootstrap_features=True,
    random_state=42
)

patches.fit(X_train, y_train)
print("Random Patches Accuracy:", accuracy_score(y_test, patches.predict(X_test)))
```

```python
oob = BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=30,
    max_samples=0.5,
    oob_score=True,
    random_state=42
)

oob.fit(X_train, y_train)
print("OOB Score:", oob.oob_score_)
print("Test Accuracy:", accuracy_score(y_test, oob.predict(X_test)))
```
